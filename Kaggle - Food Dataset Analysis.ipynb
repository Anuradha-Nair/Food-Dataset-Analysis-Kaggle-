### EDA and Visualizations

#### 3.1: Demographics

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set style
sns.set(style="whitegrid")

# Plot: Income Distribution
plt.figure(figsize=(10, 5))
sns.histplot(df["Income"], bins=30, kde=True, color='teal')
plt.title("Income Distribution")
plt.xlabel("Income")
plt.ylabel("Count")
plt.tight_layout()
plt.show()

# Plot: Age Distribution
plt.figure(figsize=(10, 5))
sns.histplot(df["Age"], bins=30, kde=True, color='slateblue')
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Count")
plt.tight_layout()
plt.show()

# Plot: Number of Children
plt.figure(figsize=(8, 4))
sns.histplot(df["Kidhome"] + df["Teenhome"], bins=5, discrete=True, color='orange')
plt.title("Total Number of Children in Household")
plt.xlabel("Number of Children")
plt.ylabel("Count")
plt.tight_layout()
plt.show()
```

#### 🔍 EDA Insights: Demographics

    * Income Distribution:

            Most customers earn between €20,000 and €80,000.

            There’s a right-skew indicating a few high-income customers.

    * Age Distribution:

            Majority fall within the 30–60 years range.

            A peak appears in the 40s–50s, suggesting mid-career professionals.

    * Children in Household:

            Most families have 0–1 children.

            Very few have more than 2 children.



#### 💡 Business Insights from Demographics

    * Targeted Promotions:

            Focus high-end products on customers with higher income (top 20% earners).

            Use digital marketing to reach middle-aged customers (30–60) who dominate the base.

    * Child-related Campaigns:

            Products like snacks or toys might have limited appeal given most customers have 0–1 kids.

#### 3.2: Purchasing Behavior

```python
# List of purchase columns
purchase_cols = [
    "MntWines", "MntFruits", "MntMeatProducts",
    "MntFishProducts", "MntSweetProducts", "MntGoldProds"
]

# Boxplot of product spending
plt.figure(figsize=(12, 6))
df[purchase_cols].plot(kind='box', figsize=(12, 6), patch_artist=True,
                       boxprops=dict(facecolor="skyblue"))
plt.title("Spending Distribution on Product Categories")
plt.ylabel("Amount Spent")
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()

# Total spending histogram
plt.figure(figsize=(10, 5))
sns.histplot(df["MntTotal"], bins=30, kde=True, color='green')
plt.title("Distribution of Total Spending")
plt.xlabel("Total Amount Spent")
plt.ylabel("Count")
plt.tight_layout()
plt.show()
```

#### 🔍 EDA Insights: Purchasing Behavior

    * Product-wise Spending (Boxplot):

            Wines and Meat Products are the top two categories in terms of median spending.

            Gold Products have a few high spenders (outliers), but most spend very little.

    * Total Spending:

            Most customers spend under €1000 in total.

            Spending has a long tail — a few customers spend significantly more than the rest.


#### 💡 Business Insights from Purchasing Behavior

    * Product Bundling:

            Create combos of wine + meat for cross-selling since they’re the most commonly purchased together.

    * Loyalty Programs:

            Target high spenders (top 10%) with VIP benefits to retain them.

    * Gold Product Promotion:

            Needs awareness campaigns — some users engage deeply, but most don’t.

### 3.3: Campaign Responses and 3.4: Engagement Channels

```python
# Campaign acceptance rates
campaign_cols = ["AcceptedCmp1", "AcceptedCmp2", "AcceptedCmp3", "AcceptedCmp4", "AcceptedCmp5"]
campaign_response_rates = df[campaign_cols].mean()

# Barplot of campaign response
plt.figure(figsize=(10, 5))
sns.barplot(x=campaign_response_rates.index, y=campaign_response_rates.values, palette='coolwarm')
plt.title("Campaign Acceptance Rates")
plt.ylabel("Acceptance Rate")
plt.xlabel("Campaign")
plt.ylim(0, 0.2)
plt.tight_layout()
plt.show()

# Plot: Purchase channel activity
channel_cols = ["NumDealsPurchases", "NumWebPurchases", "NumCatalogPurchases", "NumStorePurchases"]
plt.figure(figsize=(10, 6))
sns.boxplot(data=df[channel_cols], palette='pastel')
plt.title("Customer Engagement Across Purchase Channels")
plt.ylabel("Number of Purchases")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

#### 🔍 EDA Insights: Campaigns & Channel Engagement

    * Campaign Acceptance:

            All campaigns have low acceptance rates (< 20%).

            Slightly better performance from Campaign 3 and 5.

    * Purchase Channels:

            Store purchases dominate, indicating a strong in-person buying preference.

            Web and catalog purchases are more variable but still active.

            Deal-based purchases occur less frequently.


#### 💡 Business Insights from Campaign & Channels

    * Campaign Optimization:

            Review creative/content for campaigns 1–2, which have particularly low performance.

            Retarget customers who accepted past campaigns (e.g., 3 & 5) for upselling.

    * Omnichannel Strategy:

            Physical store should remain a priority channel.

            Boost online shopping with exclusive web-only offers or loyalty points.

### 3.5: Behavioral Factors like recency, complaints, and overall segmentation recommendations. ​

#### 🔍 EDA Insights: Behavior

    * Recency:

            Most customers made purchases within the last 0–50 days, but a long tail shows some haven’t shopped in a while.

            Indicates opportunity for reactivation campaigns.

    * Customer Complaints:

            Only about 1.5% of customers have filed complaints.

            Suggests good customer satisfaction, or possibly underreporting.



```python
# Plot: Recency
plt.figure(figsize=(10, 5))
sns.histplot(df["Recency"], bins=30, kde=True, color='coral')
plt.title("Recency of Last Purchase (Days Ago)")
plt.xlabel("Days Since Last Purchase")
plt.ylabel("Number of Customers")
plt.tight_layout()
plt.show()

# Complaint rate
complaint_rate = df["Complain"].value_counts(normalize=True)

# Pie chart for complaint
plt.figure(figsize=(6, 6))
plt.pie(complaint_rate, labels=["No Complaint", "Complained"], autopct='%1.1f%%', colors=["lightgreen", "red"])
plt.title("Customer Complaint Rate")
plt.tight_layout()
plt.show()
```
